```python
import torch
import kagglehub
import os
from ultralytics import YOLO
import shutil
torch.set_float32_matmul_precision('high')
device = "cuda" if torch.cuda.is_available() else 'cpu'
```

```python
path = kagglehub.dataset_download("pkdarabi/bone-fracture-detection-computer-vision-project")
print(os.listdir(path+'/BoneFractureYolo8'))
data = path+'/BoneFractureYolo8/data.yaml'
files_to_save = ['results.png', 'results.csv', 'confusion_matrix.png', 'val_batch0_pred.jpg', 'args.yaml']
workdir = os.getcwd()
os.makedirs(workdir+"/docs", exist_ok=True)
```

```python
light_model = YOLO('yolov8n.pt')
results = light_model.train(data=data, epochs=50, imgsz=640, batch=64, 
    mosaic=0.0, # dont forget to remove augmentations
    mixup=0.0,
    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.4,
    fliplr=0.5,
    degrees=10.0,
    single_cls=True, # only-fracture feature
)
```

```python
for i in files_to_save:
    shutil.copy(workdir+"/runs/detect/train/"+i, workdir+"/docs/fast_"+i)
shutil.copy(workdir+'/runs/detect/train/weights/best.pt', workdir+"/model_fast.pt")
```

```python
model = YOLO('yolov8m.pt')
results = model.train(data=data, epochs=25, imgsz=640, batch=64,
    mosaic=0.0, # dont forget to remove augmentations
    mixup=0.0,
    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.4,
    fliplr=0.5,
    degrees=10.0,
    #single_cls=True, # only-fracture feature
)
```

```python
for i in files_to_save:
    shutil.copy(workdir+"/runs/detect/train/"+i, workdir+"/docs/precise_"+i)
shutil.copy(workdir+'/runs/detect/train/train/weights/best.pt', workdir+"/model_precise.pt")
```

```python
from IPython.display import FileLink
shutil.make_archive(workdir+'/metrics', 'zip', workdir+'/docs')
display(FileLink('model_precise.pt'))
display(FileLink('model_fast.pt'))
display(FileLink('metrics.zip'))
```
