```python
import kagglehub
import os
from ultralytics import YOLO
import shutil
```

```python
path = kagglehub.dataset_download("orvile/human-bone-fractures-image-dataset-hbfmid")
print(os.listdir(path))
files_to_save = ['results.png', 'results.csv', 'confusion_matrix.png', 'val_batch0_pred.jpg', 'args.yaml']
os.makedirs("../docs", exist_ok=True)
os.makedirs("../weights", exist_ok=True)
```

```python
light_model = YOLO('yolov8n.pt')
results = light_model.train(data=path+'/Human Bone Fractures Multi-modal Image Dataset (HBFMID)/Bone Fractures Detection/data.yaml', epochs=100, patience=5, imgsz=640)
```

```python
for i in files_to_save:
    shutil.copy("runs/detect/train/"+i, "../docs/fast_"+i)
shutil.copy('runs/detect/train/weights/best.pt', "../weights/model_fast.pt")
```

```python
model = YOLO('yolov8s.pt')
results = model.train(data=path+'/Human Bone Fractures Multi-modal Image Dataset (HBFMID)/Bone Fractures Detection/data.yaml', epochs=100, patience=7, imgsz=640)
```

```python
for i in files_to_save:
    shutil.copy("runs/detect/train-2/"+i, "../docs/precise_"+i)
shutil.copy('runs/detect/train-2/weights/best.pt', "../weights/model_precise.pt")
```

```python
from IPython.display import FileLink
shutil.make_archive('../docs', 'zip', '../docs')
display(FileLink('../weights/model_precise.pt'))
display(FileLink('../weights/model_fast.pt'))
display(FileLink('../docs.zip'))
```
